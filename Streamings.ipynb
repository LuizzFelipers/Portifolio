```python
import pandas as pd
import plotly.express as px
```

```python
netflix = pd.read_excel('/content/sample_data/STREAMING.xlsx', sheet_name='NETFLIX')

hbo = pd.read_excel('/content/sample_data/STREAMING.xlsx', sheet_name='HBO')

disney = pd.read_excel('/content/sample_data/STREAMING.xlsx', sheet_name='DISNEY')

amazon = pd.read_excel('/content/sample_data/STREAMING.xlsx', sheet_name='AMAZON_PRIME')
```

```python
hbo = hbo.dropna()
```

```python
netflix = netflix.rename(columns={"REVENUE":"REVENUE(US$ BN)","SUBSCRIBERS":"SUBSCRIBERS(MM)"})
hbo = hbo.rename(columns={"REVENUE":"REVENUE(US$ BN)","SUBSCRIBERS":"SUBSCRIBERS(MM)"})
disney = disney.rename(columns={"REVENUE":"REVENUE(US$ BN)","SUBSCRIBERS":"SUBSCRIBERS(MM)"})
amazon = amazon.rename(columns={"REVENUE":"REVENUE(US$ BN)","SUBSCRIBERS":"SUBSCRIBERS(MM)"})
```

```python
fig_netflix = px.line(
    netflix,
    x="YEAR",
    y="REVENUE(US$ BN)",
    title="Netflix - Revenue(US$ BN) Year",
    markers="o",
    text="REVENUE(US$ BN)",
    template="plotly_dark",
    color_discrete_sequence=["red"]
)

fig_netflix.update_traces(textposition="top center")

fig_netflix.update_yaxes(tickprefix="US$ ", tickformat=",")

fig_netflix.update_traces(
    hovertemplate="%{x}<br>US$ %{y:,.0f}<extra></extra>"
)

fig_netflix.update_traces(line_shape="spline")
fig_netflix.update_traces(line=dict(width=8))
fig_netflix.update_traces(marker=dict(size=20))

fig_netflix.update_xaxes(showgrid=False)
fig_netflix.update_yaxes(showgrid=False)

fig_netflix.show()
```

```python
fig_netflix = px.line(
    netflix,
    x="YEAR",
    y="SUBSCRIBERS(MM)",
    title="Netflix - Subscribers",
    markers="o",
    text="SUBSCRIBERS(MM)",
    template="plotly_dark",
    color_discrete_sequence=["red"]
)

fig_netflix.update_traces(line_shape="spline")
fig_netflix.update_traces(line=dict(width=8))
fig_netflix.update_traces(marker=dict(size=20))

fig_netflix.update_traces(textposition="top center")

fig_netflix.update_traces(line_shape="spline")

fig_netflix.update_xaxes(showgrid=False)
fig_netflix.update_yaxes(showgrid=False)

fig_netflix.show()
```

```python
sheets = pd.read_excel("/content/sample_data/STREAMING.xlsx", sheet_name=None)

df = pd.concat(
    [d.assign(empresa=nome) for nome, d in sheets.items()],
    ignore_index=True,
)

print(df["empresa"].value_counts())
print(df.head())
```

```python
fig = px.line(
    df,
    x="YEAR",
    y="SUBSCRIBERS",
    color="empresa",
    color_discrete_map={
        "NETFLIX":       "#E50914",
        "HBO":           "#8A2BE2",
        "AMAZON_PRIME":  "#00A8E1",
        "DISNEY":        "#113CCF",
    },
    markers=True,
    template="plotly_dark",
    title="Assinantes por Ano (MM)",
)
fig.update_xaxes(showgrid=False, dtick=1)
fig.update_yaxes(showgrid=False)
fig.show()
```

```python
fig = px.line(
    df,
    x="YEAR",
    y="REVENUE",
    color="empresa",
    color_discrete_map={
        "NETFLIX":       "#E50914",
        "HBO":           "#8A2BE2",
        "AMAZON_PRIME":  "#00A8E1",
        "DISNEY":        "#113CCF",
    },
    markers=True,
    template="plotly_dark",
    title="Receita Anual(US$ Bn)",
)
fig.update_xaxes(showgrid=False, dtick=1)
fig.update_yaxes(showgrid=False)
fig.show()
```

```python
df["SUBSCRIBERS"] = pd.to_numeric(df["SUBSCRIBERS"], errors='coerce')
df["var_assinantes"] = df.groupby("empresa")["SUBSCRIBERS"].pct_change() * 100
df["var_receita"]    = df.groupby("empresa")["REVENUE"].pct_change() * 100
```

```python
fig_var = px.bar(
    df,
    x="YEAR",
    y="var_assinantes",
    color="empresa",
    color_discrete_map={
        "NETFLIX":       "#E50914",
        "HBO":           "#8A2BE2",
        "AMAZON_PRIME":  "#00A8E1",
        "DISNEY":        "#113CCF",
    },
    template="plotly_dark",
    title="Variação(%) dos Assinantes Anual"
)

fig_var.update_xaxes(showgrid=False, dtick=1)
fig_var.update_yaxes(showgrid=False)
fig_var.show()
```

```python
fig_var = px.bar(
    df,
    x="YEAR",
    y="var_receita",
    color="empresa",
    color_discrete_map={
        "NETFLIX":       "#E50914",
        "HBO":           "#8A2BE2",
        "AMAZON_PRIME":  "#00A8E1",
        "DISNEY":        "#113CCF",
    },
    template="plotly_dark",
    title="Variação(%) dos Assinantes Anual",
    labels={"YEAR":"Ano", "var_receita":"Variação(%) da Receita"}
)

fig_var.update_xaxes(showgrid=False, dtick=1)
fig_var.update_yaxes(showgrid=False)
fig_var.show()
```

```python
df["base_sub"] = df.groupby("empresa")["SUBSCRIBERS"].transform("first")
df["base_rec"] = df.groupby("empresa")["REVENUE"].transform("first")

df["acum_assinantes"] = (df["SUBSCRIBERS"] / df["base_sub"] - 1) * 100
df["acum_receita"]    = (df["REVENUE"]    / df["base_rec"] - 1) * 100
```

```python
df
```

```python
fig_var = px.line(
    df,
    x="YEAR",
    y="acum_assinantes",
    title="Crescimento acumulado de Assinantes",
    color="empresa",
    color_discrete_map={
        "NETFLIX":       "#E50914",
        "HBO":           "#8A2BE2",
        "AMAZON_PRIME":  "#00A8E1",
        "DISNEY":        "#113CCF",
    },
    template="plotly_dark"
)

fig_var.update_xaxes(showgrid=False, dtick=1)
fig_var.update_yaxes(showgrid=False)

fig_var.show()
```

```python
fig_var = px.line(
    df,
    x="YEAR",
    y="acum_receita",
    title="Crescimento acumulado da Receita",
    color="empresa",
    color_discrete_map={
        "NETFLIX":       "#E50914",
        "HBO":           "#8A2BE2",
        "AMAZON_PRIME":  "#00A8E1",
        "DISNEY":        "#113CCF",
    },
    template="plotly_dark"
)

fig_var.update_xaxes(showgrid=False, dtick=1)
fig_var.update_yaxes(showgrid=False)

fig_var.show()
```
